```python
import pandas as pd
lit = pd.read_csv(r"C:\Users\nmqui\Downloads\municipal_litigation_finances_2000_2025.csv")
print(lit.shape)
print(list(lit.columns))
```

```python
for c in lit.columns:
    print(c)
```

```python
# Find the defendant case-count column (not a rate)
def_cols = [c for c in lit.columns if "defendant" in c.lower()]
count_col = [c for c in def_cols if not any(w in c.lower() for w in ["rate", "per", "10k", "share", "pct"])]
print("Defendant columns found:", def_cols)
print("Using:", count_col[0])
count_col = count_col[0]

gateway = ["Attleboro","Barnstable","Brockton","Chelsea","Chicopee","Everett","Fall River","Fitchburg",
           "Haverhill","Holyoke","Lawrence","Leominster","Lowell","Lynn","Malden","Methuen","New Bedford",
           "Peabody","Pittsfield","Quincy","Revere","Salem","Springfield","Taunton","Westfield","Worcester"]

d = lit[(lit["year"] >= 2006) & (lit["year"] <= 2024)].copy()
d["group"] = d["municipality"].isin(gateway).map({True: "Gateway Cities", False: "Rest of state"})

trend = d.groupby(["year", "group"])[[count_col, "population"]].sum().reset_index()
trend["defendant_cases_per_10k"] = (trend[count_col] / trend["population"] * 10000).round(2)
print(trend.pivot(index="year", columns="group", values="defendant_cases_per_10k"))
```

```python
trend[["year", "group", "defendant_cases_per_10k"]].to_csv(
    r"C:\Users\nmqui\Downloads\gateway_vs_state_defendant_rate_2006_2024.csv", index=False)
print("saved")
```
